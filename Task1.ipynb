```python
import pandas as pd
import matplotlib.pyplot as plt


file_path = "/content/drive/MyDrive/2nd/task 1/API_SP.POP.TOTL_DS2_en_csv_v2_2590.csv"
df = pd.read_csv(file_path, skiprows=4)

population_2023 = df["2023"].dropna()


plt.figure(figsize=(10, 6))
plt.hist(population_2023, bins=30, color='skyblue', edgecolor='black')
plt.title('Distribution of Total Population by Country (2023)')
plt.xlabel('Population')
plt.ylabel('Number of Countries')
plt.grid(True)
plt.tight_layout()
plt.show()
```
